# Financial Fraud Detection & Risk Intelligence System

## 01 — Data Audit

### Objective

Understand the structure, quality, completeness, and reliability
of the financial transaction dataset before exploratory data
analysis and machine learning.

### Dataset

Financial Fraud Detection Dataset

### Target Variable

Fraudulent

- 0 = Non-fraud
- 1 = Fraud

```python
import pandas as pd
import numpy as np
```

```python
# Load raw dataset

df = pd.read_csv(
    "../data/raw/financial_fraud_detection_dataset.csv"
)

print("Dataset loaded successfully")
```

```python
# Dataset shape

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# First five rows
df.head()

# Last five rows
df.tail()

# Random sample
df.sample(5, random_state=42)
```

```python
# Dataset information
df.info()

# Column names
print(df.columns.tolist())
```

```python
# Missing-value summary

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

print(missing_summary)
```

```python
# Complete duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Duplicate transaction IDs
print(
    "Duplicate Transaction IDs:",
    df["Transaction_ID"].duplicated().sum()
)
```

```python
# Unique values per column

unique_summary = pd.DataFrame({
    "Column": df.columns,
    "Unique_Values": df.nunique().values
})

print(unique_summary)
```

```python
# Data types
print(df.dtypes)
```

```python
# Target distribution

target_summary = (
    df["Fraudulent"]
    .value_counts()
    .sort_index()
    .reset_index()
)

target_summary.columns = ["Fraudulent", "Count"]

target_summary["Percentage"] = (
    target_summary["Count"] / len(df) * 100
).round(2)

print(target_summary)
```

```python
# Identifier uniqueness

identifier_columns = [
    "Transaction_ID",
    "Customer_ID"
]

for column in identifier_columns:
    print(
        column,
        "Unique values:",
        df[column].nunique(),
        "Total rows:",
        len(df)
    )
```

```python
# Numerical feature summary

print(df.describe())
```

```python
# Categorical columns

categorical_columns = [
    "Merchant_Category",
    "Payment_Method",
    "Device_Type",
    "Location",
    "Is_International",
    "Suspicious_Keyword"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False))
```

```python
# Convert date for validation

df["Transaction_Date"] = pd.to_datetime(
    df["Transaction_Date"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

print("Invalid dates:", df["Transaction_Date"].isna().sum())

print("Minimum date:", df["Transaction_Date"].min())
print("Maximum date:", df["Transaction_Date"].max())
```

## Data Audit — Key Findings

1. The dataset contains 5,000 transaction records.
2. The target variable is Fraudulent.
3. The dataset contains 482 fraud records and 4,518 non-fraud records.
4. No missing values were detected during the audit.
5. No complete duplicate rows were detected.
6. Transaction_ID is a unique identifier.
7. Customer_ID is repeated across some transactions.
8. Transaction_Date requires date parsing using the expected format.
9. Suspicious_Keyword requires a separate leakage review.
10. The dataset should be documented as synthetic.
